```python
def brier_score(probabilities,outcomes):
    if len(probabilities)!=len(outcomes) or not probabilities: raise ValueError('equal non-empty arrays required')
    return sum((float(p)-int(y))**2 for p,y in zip(probabilities,outcomes))/len(outcomes)
print('Helper loaded; verified settled data is required for metrics.')
```
